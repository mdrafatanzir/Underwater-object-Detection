```python
!pip install -q ultralytics opencv-python numpy matplotlib

import os, cv2, glob, json, time, random
import numpy as np
import torch
import matplotlib.pyplot as plt
from ultralytics import RTDETR

# =========================
# Reproducibility
# =========================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# =========================
# Paths
# =========================
DATASET_PATH = "/kaggle/input/datasets/abidarahman1997/dataset1/Dataset1"

TEST_IMG = f"{DATASET_PATH}/test/images"
TEST_LBL = f"{DATASET_PATH}/test/labels"

RESULTS_DIR = "/kaggle/working/rtdetr_results"
os.makedirs(RESULTS_DIR, exist_ok=True)

# =========================
# Config
# =========================
CLASSES = ["propeller", "pipe_type2", "red_fin", "net", "qr_codes", "pipe"]

CONF_THRESHOLD = 0.5
IOU_THRESHOLD = 0.5
VIS_SAMPLES = 5
EPOCHS = 30
IMG_SIZE = 640
BATCH_SIZE = 16

# =========================
# Dataset YAML
# =========================
yaml_text = f"""
path: {DATASET_PATH}
train: train/images
val: valid/images
test: test/images

nc: 6
names:
  - propeller
  - pipe_type2
  - red_fin
  - net
  - qr_codes
  - pipe
""".strip()

with open("/kaggle/working/Dataset1.yaml", "w") as f:
    f.write(yaml_text)

# =========================
# Helper functions
# =========================
def list_images(folder):
    return sorted(
        glob.glob(os.path.join(folder, "*.jpg")) +
        glob.glob(os.path.join(folder, "*.jpeg")) +
        glob.glob(os.path.join(folder, "*.png"))
    )

def read_yolo_labels(txt_file, W, H):
    boxes, labels = [], []

    if not os.path.exists(txt_file):
        return boxes, labels

    with open(txt_file, "r") as f:
        for line in f:
            parts = line.strip().split()

            if len(parts) != 5:
                continue

            c, x, y, w, h = parts
            c = int(float(c))
            x, y, w, h = map(float, [x, y, w, h])

            x1 = (x - w / 2) * W
            y1 = (y - h / 2) * H
            x2 = (x + w / 2) * W
            y2 = (y + h / 2) * H

            boxes.append([x1, y1, x2, y2])
            labels.append(c)

    return boxes, labels

def iou_xyxy(a, b):
    xA = max(a[0], b[0])
    yA = max(a[1], b[1])
    xB = min(a[2], b[2])
    yB = min(a[3], b[3])

    inter = max(0, xB - xA) * max(0, yB - yA)

    areaA = max(0, a[2] - a[0]) * max(0, a[3] - a[1])
    areaB = max(0, b[2] - b[0]) * max(0, b[3] - b[1])

    union = areaA + areaB - inter

    return inter / union if union > 0 else 0.0

# =========================
# Model
# =========================
MODEL_NAME = "RT-DETR-L"

model = RTDETR("rtdetr-l.pt")

device_id = 0 if torch.cuda.is_available() else "cpu"

param_count = sum(p.numel() for p in model.model.parameters())

print("Model:", MODEL_NAME)
print("Parameters:", param_count)
print("Device:", device_id)

# =========================
# Print only after each epoch
# =========================
epoch_start_times = {}

def on_train_epoch_start(trainer):
    epoch_start_times["start"] = time.time()

def on_train_epoch_end(trainer):
    epoch_time = time.time() - epoch_start_times.get("start", time.time())

    epoch = trainer.epoch + 1
    total_epochs = trainer.epochs

    loss_items = trainer.loss_items

    if loss_items is not None:
        try:
            loss_text = " | ".join(
                [f"loss_{i}: {float(v):.4f}" for i, v in enumerate(loss_items)]
            )
        except:
            loss_text = str(loss_items)
    else:
        loss_text = "loss not available"

    print(
        f"Epoch {epoch}/{total_epochs} completed | "
        f"{loss_text} | "
        f"Time: {epoch_time:.2f} sec"
    )

model.add_callback("on_train_epoch_start", on_train_epoch_start)
model.add_callback("on_train_epoch_end", on_train_epoch_end)

# =========================
# Training
# =========================
train_start = time.time()

model.train(
    data="/kaggle/working/Dataset1.yaml",
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=BATCH_SIZE,
    device=device_id,
    project=RESULTS_DIR,
    name="train",
    exist_ok=True,
    verbose=False,
    plots=False
)

train_end = time.time()
training_seconds = train_end - train_start

print(f"\nTotal training time: {training_seconds:.2f} seconds")

# =========================
# Validation on test split
# =========================
metrics = model.val(
    data="/kaggle/working/Dataset1.yaml",
    split="test",
    project=RESULTS_DIR,
    name="val",
    exist_ok=True,
    verbose=False,
    plots=False
)

# =========================
# Manual evaluation
# =========================
test_images = list_images(TEST_IMG)

times = []
TP = 0
FP = 0
FN = 0
correct = 0
total_preds = 0

for img_path in test_images:
    img = cv2.imread(img_path)

    if img is None:
        continue

    H, W, _ = img.shape

    start = time.time()
    preds = model(img, verbose=False)[0]
    end = time.time()

    times.append(end - start)

    label_path = os.path.join(
        TEST_LBL,
        os.path.splitext(os.path.basename(img_path))[0] + ".txt"
    )

    gt_boxes, gt_labels = read_yolo_labels(label_path, W, H)
    gt = list(zip(gt_boxes, gt_labels))

    matched_gt = set()

    for box, cls, conf in zip(preds.boxes.xyxy, preds.boxes.cls, preds.boxes.conf):
        if float(conf) < CONF_THRESHOLD:
            continue

        pb = box.cpu().numpy().tolist()
        pl = int(cls.cpu().item())

        total_preds += 1

        best_iou = 0.0
        best_idx = -1

        for j, (gb, gl) in enumerate(gt):
            if j in matched_gt or pl != gl:
                continue

            current_iou = iou_xyxy(pb, gb)

            if current_iou > best_iou:
                best_iou = current_iou
                best_idx = j

        if best_iou >= IOU_THRESHOLD and best_idx != -1:
            TP += 1
            correct += 1
            matched_gt.add(best_idx)
        else:
            FP += 1

    FN += len(gt) - len(matched_gt)

precision = (TP / (TP + FP) * 100) if (TP + FP) > 0 else 0.0
recall = (TP / (TP + FN) * 100) if (TP + FN) > 0 else 0.0
detection_accuracy = (correct / total_preds * 100) if total_preds > 0 else 0.0

avg_inference_time = float(np.mean(times)) if times else 0.0
total_inference_time = float(np.sum(times)) if times else 0.0
fps = (len(test_images) / total_inference_time) if total_inference_time > 0 else 0.0

# =========================
# Visualization
# =========================
for img_path in test_images[:VIS_SAMPLES]:
    img = cv2.imread(img_path)

    if img is None:
        continue

    H, W, _ = img.shape
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    label_path = os.path.join(
        TEST_LBL,
        os.path.splitext(os.path.basename(img_path))[0] + ".txt"
    )

    gt_boxes, gt_labels = read_yolo_labels(label_path, W, H)

    for b, lbl in zip(gt_boxes, gt_labels):
        x1, y1, x2, y2 = map(int, b)

        if 0 <= lbl < len(CLASSES):
            class_name = CLASSES[lbl]
        else:
            class_name = "unknown"

        cv2.rectangle(img_rgb, (x1, y1), (x2, y2), (0, 0, 255), 2)
        cv2.putText(
            img_rgb,
            f"GT: {class_name}",
            (x1, max(20, y1 - 5)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (0, 0, 255),
            1
        )

    detected_names = []

    pred = model(img, verbose=False)[0]

    for box, cls, conf in zip(pred.boxes.xyxy, pred.boxes.cls, pred.boxes.conf):
        if float(conf) < CONF_THRESHOLD:
            continue

        cls_id = int(cls.cpu().item())

        if not (0 <= cls_id < len(CLASSES)):
            continue

        x1, y1, x2, y2 = map(int, box.cpu().numpy().tolist())

        class_name = CLASSES[cls_id]
        detected_names.append(class_name)

        cv2.rectangle(img_rgb, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(
            img_rgb,
            f"Pred: {class_name} {float(conf):.2f}",
            (x1, max(40, y1 - 5)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (0, 255, 0),
            1
        )

    plt.figure(figsize=(10, 10))
    plt.imshow(img_rgb)
    plt.title(
        f"Image: {os.path.basename(img_path)}\n"
        f"Detected: {', '.join(sorted(set(detected_names))) if detected_names else 'No object detected'}"
    )
    plt.axis("off")
    plt.show()

# =========================
# Summary
# =========================
summary = {
    "model": "RT-DETR-L",
    "parameters": int(param_count),
    "training_time_seconds": float(training_seconds),
    "avg_inference_time_seconds": avg_inference_time,
    "total_test_inference_time_seconds": total_inference_time,
    "fps": fps,
    "map50_percent": float(metrics.box.map50 * 100),
    "map50_95_percent": float(metrics.box.map * 100),
    "precision_percent": float(precision),
    "recall_percent": float(recall),
    "detection_accuracy_percent": float(detection_accuracy),
    "epochs": EPOCHS,
    "batch_size": BATCH_SIZE,
    "image_size": IMG_SIZE
}

with open(os.path.join(RESULTS_DIR, "summary.json"), "w") as f:
    json.dump(summary, f, indent=2)

print("\nFinal summary:")
print(json.dumps(summary, indent=2))
```
